```python
import matplotlib.pyplot as plt
subjects = ["math","science","english","history"]
students = [40,60,30,25]
plt.bar(subjects,students)
plt.title("bar chart")
plt.xlabel("subjects")
plt.ylabel("students")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
students = ["Ravi","Sai","uma"]

maths = [99,90,100]
science = [88,80,90]
hindi = [77,70,80]
x = np.arange(len(students))
width = 0.25
plt.bar(x-width,maths,width,label="maths")
plt.bar(x,science,width,label="science")
plt.bar(x+width,hindi,width,label="hindi")
plt.xlabel("students")
plt.ylabel("marks")
plt.title("bar chart")
plt.xticks(x,students)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
activities = ['study','sleep','sports','other']
hours = [5,8,3,8]
plt.pie(hours,labels=activities)
plt.title("Daily Time Allocation")
plt.show()
```
